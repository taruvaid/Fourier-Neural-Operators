```python
import numpy as np
import tensorboardX
from fbpinns.constants import Constants
from fbpinns.trainers import FBPINNTrainer, PINNTrainer
from fbpinns.schedulers import AllActiveSchedulerND
from fbpinns.problems import HarmonicOscillator1D  # Using existing problem instead of non-existent Cos1D_1
```

```python
c1 = Constants(
    run="FBPINN_HarmonicOscillator",  # run name
    problem=HarmonicOscillator1D,  # problem class
    problem_init_kwargs=dict(d=2, w0=20),  # problem parameters
    decomposition_init_kwargs=dict(
        subdomain_xs=[np.linspace(0, 1, 5)],  # defines subdomains
        subdomain_ws=[0.6*np.ones(5)],  # width of overlapping regions - must be large enough for adjacent subdomains to overlap
        unnorm=(0., 1.),  # unnormalisation
    ),
    network_init_kwargs=dict(
        layer_sizes=[1, 16, 1],  # network architecture
    ),
    n_steps=5000,  # number of training steps
    ns=((200,),),  # batch shapes for training
    n_test=(400,),  # batch shape for testing
    summary_freq=100,  # output training stats every 100 steps
    test_freq=500,  # test every 500 steps
    show_figures=False,  # don't show figures during training (set to True to see plots)
)

run = FBPINNTrainer(c1)  # train FBPINN
run.train()

c2 = Constants(
    run="PINN_HarmonicOscillator",
    problem=HarmonicOscillator1D,
    problem_init_kwargs=dict(d=2, w0=20),
    decomposition_init_kwargs=dict(
        subdomain_xs=[np.linspace(0, 1, 5)],
        subdomain_ws=[0.6*np.ones(5)],  # must be large enough for adjacent subdomains to overlap
        unnorm=(0., 1.),
    ),
    network_init_kwargs=dict(
        layer_sizes=[1, 32, 1],  # different network size
    ),
    n_steps=5000,
    ns=((200,),),
    n_test=(400,),
    summary_freq=100,
    test_freq=500,
    show_figures=False,
)

run = PINNTrainer(c2)  # train PINN
run.train()
```
