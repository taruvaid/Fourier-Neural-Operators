```python
import numpy as np
import matplotlib.pyplot as plt

# Define function
def f(x):
    return -0.5*(x-2)**2 + 3  # smooth concave-down curve

# Interval [a, b]
a, b = 0, 4

# Create x values
x = np.linspace(-1, 5, 400)
y = f(x)

# Points A and B
A = (a, f(a))
B = (b, f(b))

# Secant slope
m_secant = (f(b) - f(a)) / (b - a)

# Find point C such that f'(c) = secant slope
# For this function, derivative is f'(x) = -(x-2)
# Solve -(c-2) = m_secant → c = 2 - m_secant
c = 2 - m_secant
C = (c, f(c))

# Tangent line at C
tangent_line = m_secant * (x - c) + f(c)

# Secant line
secant_line = m_secant * (x - a) + f(a)

# Plot
plt.figure(figsize=(12,5))
plt.plot(x, y, label='f(x)')
plt.plot(x, secant_line, '--', label='Secant line AB - Average rate of change')
plt.plot(x, tangent_line, '-', label='Tangent at C - Instantaneous rate of change')

# Plot points
plt.scatter(*A)
plt.scatter(*B)
plt.scatter(*C)

# Labels
plt.text(A[0], A[1]-0.3, 'A')
plt.text(B[0], B[1]-0.3, 'B')
plt.text(C[0], C[1]-0.2, 'C')

# Styling
plt.axhline(0)
plt.axvline(0)
plt.legend()
plt.title("Mean Value Theorem")
plt.xlabel("x")
plt.ylabel("f(x)")


plt.show()
```

```python
import matplotlib.pyplot as plt

# Create figure
fig, ax = plt.subplots(figsize=(10, 2))

# Number line
ax.hlines(0, 0, 6, color='black')

# Set S = (1, 4]
ax.plot([1, 4], [0, 0], color='blue', linewidth=6, label='Set S = (1,4]')

# Open circle at 1 (not included)
ax.plot(1, 0, 'wo', markeredgecolor='blue', markersize=10, markeredgewidth=2)

# Closed circle at 4 (included)
ax.plot(4, 0, 'bo', markersize=10)

# Infimum label
ax.text(1, 0.15, 'infimum = 1', ha='center', color='green')
ax.plot(1, -0.2, 'go')

# Supremum label
ax.text(4, 0.15, 'supremum = 4', ha='center', color='red')
ax.plot(4, -0.2, 'ro')

# Lower bound (example: 0)
ax.text(-1, -0.5, 'lower bound (e.g. 0)', color='black')
ax.plot(0, -0.2, 'ko')

# Upper bound (example: 6)
ax.text(5.5, -0.5, 'upper bound (e.g. 6)', color='black')
ax.plot(6, -0.2, 'ko')

# Formatting
ax.set_ylim(-1, 1)
ax.set_xlim(-1, 7)
ax.set_yticks([])
ax.set_xticks(range(0, 7))
ax.spines[['left', 'right', 'top']].set_visible(False)

plt.title("Upper Bound, Lower Bound, Supremum, Infimum")
plt.show()
```
